```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def f(x):
    return (x+5)**2
```

```python
def df(x):
    return 2*(x+5)
    
```

```python
x = 2
learning_rate = 0.1
iterations = 1000
```

```python
history = [x]
```

```python
for i in range(iterations):
    gradient = df(x)

    x = x- learning_rate* gradient

    history.append(x)

    print("Iteration", i+1, "x=", round(x,4),
           "y=", round(f(x),4))
```

```python
print("Local Minima value is at ",x)
print("Local Minima is ",f(x))
```

```python
plt.plot(history, [f(val) for val in history],
         marker='o', color='red',
         markerfacecolor='blue',markeredgecolor="blue")
plt.title("Gradient Descend ")
plt.xlabel("x_val")
plt.ylabel("Func val")
plt.legend()
plt.grid()
plt.show()
```
